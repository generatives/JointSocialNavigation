```python
import numpy as np
import time
from mcts.decoupled_mcts import MCTS
from simulator.mcts_game_state import MCTSGameState
from simulator.mcts_game_state import navigation_rollout
from mcts.decoupled_mcts import MCTSConfig
from simulator.mcts_game_state import MCTSGameStateConfig
from simulator.scenario_map import ScenarioMap
```

```python
scenario = ScenarioMap.build_empty()

positions = np.array([
    [0.0, 0.0]
])
orientations = np.array([
    0.0
])
linear_velocities = np.array([
    1.0
])
angular_velocities = np.array([
    0.0
])
goal_positions = np.array([
    [1.0, 0.0]
])
starting_distances = np.linalg.norm(goal_positions - positions, axis=1)

num_agents = positions.shape[0]

mcts_config = MCTSConfig(
    num_actors=num_agents,
    rng=np.random.default_rng(),
    max_actions=[8],
)
state_config = MCTSGameStateConfig(
    mcts_config=mcts_config,
    dt=0.5,
    robot_speed=1.0,
    robot_max_linear_accel=1.0,
    robot_angular_velocity=1.0,
    robot_max_angular_accel=1.0,
    robot_radius=0.5,
    human_radius=0.5,
    uncomfortable_distance=1.75,
    map=scenario,
    starting_distances=starting_distances,
)

mcts = MCTS(mcts_config, navigation_rollout)

root_state = MCTSGameState(
    positions=positions,
    linear_velocities=linear_velocities,
    orientations=orientations,
    angular_velocities=angular_velocities,
    agent_goal_positions=goal_positions,
    accumulated_value=None,
    config=state_config,
    depth=0
)
```

```python
actions =[
    ((1.0, 0.0), 1.0)
]
child_state = root_state.apply_actions(actions)
child_state.positions
```
